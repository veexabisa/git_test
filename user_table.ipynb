### Exercise 1 :
Create a table called users with id, name, and email

```python
import sqlite3
```

```python
connection = sqlite3.connect('users_db')
cursor = connection.cursor()
```

```python
cursor.execute('''
               CREATE TABLE IF NOT EXISTS users(
               id INTEGAR PRIMARY KEY,
               name TEXT NOT NULL,
               email_address TEXT UNIQUE NOT NULL)
               '''
)
connection.commit()
```

```python
cursor.execute('''
               INSERT INTO users(id,name,email_address)
               VALUES(12,'Charlie','charlie@gmail.com'),
               (13,'Bob','bobs@yahoo.com'),
               (28,'Vicky','vicks@gmail.com'),
               (11,'Sam','samon@email.com')
               
              ''' )

connection.commit()
```
